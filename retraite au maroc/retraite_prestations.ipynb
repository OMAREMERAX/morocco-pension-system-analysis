```python
import pandas as pd
import numpy as np

# 1. Lire le fichier
path = "retraite_prestations_2012-2024.xlsx"

df = pd.read_excel(
    path,
    engine="openpyxl"
)

# 2. Renommer la première colonne
df = df.rename(columns={
    "En millions de dirhams": "regime"
})

# 3. Supprimer les lignes complètement vides
df = df.dropna(how="all")

# 4. Garder uniquement les régimes avec des données
df = df[df["regime"].notna()]

# 5. Transformer les années en lignes
df_long = df.melt(
    id_vars="regime",
    var_name="annee",
    value_name="prestations"
)

# 6. Convertir les types
df_long["annee"] = pd.to_numeric(
    df_long["annee"],
    errors="coerce"
)

df_long["prestations"] = pd.to_numeric(
    df_long["prestations"],
    errors="coerce"
)

# 7. Supprimer les lignes invalides
df_long = df_long.dropna(
    subset=["annee", "prestations"]
)

# 8. Année en entier
df_long["annee"] = df_long["annee"].astype(int)

# 9. Trier
df_long = df_long.sort_values(
    ["annee", "regime"]
).reset_index(drop=True)

# 10. Afficher
display(df_long)
```

```python
print("Dimensions :", df_long.shape)

print("\nColonnes :")
print(df_long.columns.tolist())

print("\nValeurs manquantes :")
print(df_long.isna().sum())

print("\nTypes :")
print(df_long.dtypes)
```

```python
df_long.to_csv(
    "prestations_clean.csv",
    index=False,
    encoding="utf-8-sig"
)

print("✅ Prestations nettoyé et enregistré")
```
