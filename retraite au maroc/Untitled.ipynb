```python

import pandas as pd
from pathlib import Path

folder = Path("data cleaning")

years = list(range(2012, 2025))

data = {
    "RETRAITES_base": [
        533938, 560868, 600707, 642741, 686408,
        722878, 768081, 800912, 843536, 887308,
        929703, 975000, 1023753
    ],
    "RETRAITES_complementaires": [
        101966, 107834, 112475, 117476, 122447,
        128238, 132781, 138154, 143053, 147729,
        152669, 158023, 163195
    ],
    "CONJOINTS_base": [
        245628, 251353, 264763, 267781, 273627,
        291506, 298883, 312108, 337008, 353538,
        356698, 376027, 387098
    ],
    "CONJOINTS_complementaires": [
        43471, 44160, 42817, 44299, 45739, 47370,
        47871, 49286, 51104, 53543, 55012, 56282, 56071
    ],
    "ORPHELINS_base": [
        29742, 28069, 25896, 26532, 24903,
        25062, 24611, 58662, 61069, 70926,
        69467, 69730, 71826
    ],
    "ORPHELINS_complementaires": [
        391, 398, 394, 453, 486, 515, 558,
        640, 688, 725, 968, 1021, 1071
    ]
}

rows = []

for key, values in data.items():
    categorie, regime = key.split("_", 1)
    regime = (
        "Régimes de base"
        if regime == "base"
        else "Régimes complémentaires"
    )

    for annee, nombre in zip(years, values):
        rows.append({
            "annee": annee,
            "regime": regime,
            "categorie": categorie,
            "beneficiaires": nombre
        })

df_beneficiaires = pd.DataFrame(rows)

# Contrôle : 13 années x 2 régimes x 3 catégories
print("Dimensions :", df_beneficiaires.shape)
print(df_beneficiaires.head(10))

# Sauvegarde dans un nouveau fichier
output = folder / "beneficiaires_clean.csv"
df_beneficiaires.to_csv(output, index=False, encoding="utf-8-sig")

print("\nFichier créé :", output)
```

```python

# Total des bénéficiaires par année
totaux = (
    df_beneficiaires
    .groupby("annee")["beneficiaires"]
    .sum()
    .reset_index(name="total_calcule")
)

print(totaux.to_string(index=False))

# Vérifier les doublons
print(
    "\nDoublons année-régime-catégorie :",
    df_beneficiaires.duplicated(
        subset=["annee", "regime", "categorie"]
    ).sum()
)

# Vérifier les valeurs manquantes
print("\nValeurs manquantes :")
print(df_beneficiaires.isna().sum())
```

```python

import pandas as pd
from pathlib import Path

dossier = Path("data cleaning")

df_actifs = pd.read_csv(dossier / "actifs_cotisants_clean.csv")
df_cotisations = pd.read_csv(dossier / "retraite_cotisations_clean.csv")

cles = ["regime", "annee"]

comparaison = df_actifs.merge(
    df_cotisations,
    on=cles,
    how="outer",
    suffixes=("_actifs", "_cotisations"),
    indicator=True
)

comparaison["valeurs_identiques"] = (
    comparaison["actifs_cotisants_actifs"]
    == comparaison["actifs_cotisants_cotisations"]
)

print("Nombre de lignes comparées :", len(comparaison))
print("Lignes présentes dans les deux fichiers :",
      (comparaison["_merge"] == "both").sum())
print("Valeurs identiques :",
      comparaison["valeurs_identiques"].sum())
print("Valeurs différentes :",
      (~comparaison["valeurs_identiques"]).sum())

print("\nExemples de comparaison :")
print(comparaison.head(10).to_string(index=False))
```
