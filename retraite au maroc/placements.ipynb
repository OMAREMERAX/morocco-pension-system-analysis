```python
import pandas as pd

path = "retraite_placements_2012-2024.xlsx"

df = pd.read_csv(
    path,
    encoding="utf-8-sig"
)

display(df.head())
print(df.shape)
print(df.columns.tolist())
```

```python
df = df.rename(columns={
    "actifs_cotisants": "placements"
})

df["annee"] = df["annee"].astype(int)
df["placements"] = pd.to_numeric(
    df["placements"],
    errors="coerce"
)

df = df.sort_values(
    ["annee", "regime"]
).reset_index(drop=True)

display(df)
```

```python
print("Dimensions :", df.shape)
print("\nValeurs manquantes :")
print(df.isna().sum())

print("\nTypes :")
print(df.dtypes)
```

```python
df.to_csv(
    "placements_clean.csv",
    index=False,
    encoding="utf-8-sig"
)

print("✅ Placements nettoyé et enregistré")
```
